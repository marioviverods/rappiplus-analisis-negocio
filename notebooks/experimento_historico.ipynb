# Experimento histórico de checkout
Código y salidas históricas del proyecto; no ejecutados en esta revisión. Requiere data/raw/experiment_checkout_ui.csv. p = 0.4161 no demuestra equivalencia.


```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd()
if not (ROOT/'data').is_dir(): ROOT = ROOT.parent
from scipy import stats
import numpy as np
 
# Cargar dataset del experimento
experiment = pd.read_csv(ROOT/'data/raw/experiment_checkout_ui.csv')
experiment.head()
 
# Explorar el dataset
print(experiment['variante'].value_counts())
print(experiment.groupby('variante')['convirtio'].mean())
```

```python
# Separar grupos
control     = experiment[experiment['variante'] == 'control']['convirtio']
tratamiento = experiment[experiment['variante'] == 'tratamiento']['convirtio']
 
# Calcular tasas de conversión
tasa_control     = control.mean()
tasa_tratamiento = tratamiento.mean()
print(f"\nTasa de conversión control:     {tasa_control:.4f} ({tasa_control*100:.2f}%)")
print(f"Tasa de conversión tratamiento: {tasa_tratamiento:.4f} ({tasa_tratamiento*100:.2f}%)")
 
# Aplicar Z-test de proporciones
n_control     = len(control)
n_tratamiento = len(tratamiento)
 
# Proporción combinada
p_pool = (control.sum() + tratamiento.sum()) / (n_control + n_tratamiento)
 
# Estadístico Z
se = np.sqrt(p_pool * (1 - p_pool) * (1/n_control + 1/n_tratamiento))
z_stat = (tasa_tratamiento - tasa_control) / se
 
# p-valor (dos colas)
p_valor = 2 * (1 - stats.norm.cdf(abs(z_stat)))
 
print(f"\nEstadístico Z: {z_stat:.4f}")
print(f"p-valor:       {p_valor:.4f}")
```
